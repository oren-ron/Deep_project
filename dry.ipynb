### Question 6

<p>
The "inductive bias" is all of the assumptions that our algorithm makes in order to operate on unknown data.
The choises we make about the model's arcitechture introduce the inductive bias on our models. <br/> <br/>
One of the inductive biases of CNNs is locality. CNNs assume that all of the important features can be detected
spatially close to each other, that is what convultions are for.
Another inductive bias is that the location of the features in the image doesn't matter, only the fact that the feature exists.
Through down sampling (like max pools) the CNN loses track of the original locations of the features and that is why it doesn't
matter to it. <br/> <br/>
Pros of inductive bias are better generalization and a smaller hypothesis space. When we introduce inductive bias we are using
the domain knowledge in order to make better predictions on unseen data. For example in images, we assume that whether the image
contains a local feature such as an eye, is very important to the classification in the end.
Also, the inductive bias helps us narrow down the options for our search. For example, we assume that a decision tree will
act very poorly on images because it should not have the capability to "understand" the images well enough. <br/>
Cons of the indcutive bias are the non-flexibility and potential overfitting. When we narrow down the hypothesis space we
limit the algorithm's flexibility, which could damage the search for the best model needed, espicially if we don't have
a lot of domain knowledge. Also we're prune to overfitting in such way that the bias we introduced will result in a specific
and poor distribution of the data that does not fit the real distribtuion.<br/> <br/>
We would want to use inductive bias when the domain is very specific, for example in images we know that CNNs would work well,
but RNNs would probably not, so the bias will help us in our learning process.
Also, when we have small dataset, the inductive bias can help us improve the learning process by making more assumptions (assuming those are correct). <br/>
We would want to avoid the inductive bias when the data is very diverse or even uknown. That way we can negate any assumptions on the data
and by using more computing power and time we could get better results.
Also, when we use data that demands learning very complex patterns, we should use more flexible models such as MLPs or transformers
in order to allow the model to learn those patterns, free from restrictive assumptions.
</p>

### Question 8

<p>
Each pixel's brightness in the attention map represents the level of connection between the each pair of english-french tokens.<br/>
Having a row with only one non-zero pixel mean that a specific token, there is only one token that attends to it. In our case, it means that the word that is translated, is translated to one specific word, and it doesn't require any other meaning from any other word in the sentence. For example, 1992 is translated perfectly to 1992 and it stands by its own without relying on any other words. <br/>
In contrast, having several non-zero pixels mean that a specific token, attends to several different tokens. Basically, the token translated gathers meaning from several tokens. For example, the word economique, gathers information from the word economic (of course) but also from the word europian, because it does matter that we're talking about the europian economy. <br/>
White pixel corresponds with a perfect attention score of 1, while gray is some score between 0 and 1. The scores are achieved by using softmax, which creates a probability distribution in proportion to all of the entries. If one entry is by far bigger than the rest, the softmax will result in ~1 in that entry and ~0 in the rest. If there are several entries close to each other, the softmax will result in a few non-zero entries and we will get those gray pixels.
</p>

The KL-divergence quantifies the difference between the approximate posterior and the true posterior and it is always non-negative due to the properties of KL-divergence. <br/>
We can't compute the DKL because we don't have the true posterior.<br/>
Since we optimize the log likelihood which is the ELBO term + the DKL and the DKL is non-negative, we get that the log likelihood is always greater than the ELBO term, where we get equality when we get to perfect case (DKL is 0). <br/>
Therefore optimizing for ELBO only, results in a lower and lower lower-bound, which in turn brings us closer to a better representation of the true data distribution. <br/> <br/>
In oreder to calculate the DKL we need the true posterior which relys on the marginal likelihood.
In order to get this we would need to calculate an infeasable integral of all posible values of the latent variables. That is, as said, infeasable, and even trying to approximate this would have taken too many computational resources for standtard training procedure. <br/> <br/>
First of all, DKL is non-negative and therefore the -DKL here could have helped us get a better lower bound, but neglecting it doesn't damage our goal. <br/>
This term measures the "distance" between the variational distribution to the prior distribution. <br/>
Ignoring this term will result in a less accurate optimization at first, but because the other terms bring us closer to our goal, this DKL term shrinks more and more and becoming more and more negligible. <br/>
So ignoring this term will make our training longer, but it doesn't damage our objective, and it becomes negligible enough to drop.

### Question 10

<p>
Bob is wrong because of a few reasons. First, the autoencoder's latent space is unstructured and unorganized at all.
Sampling a vector out of it will most likely result in a total nonsense representation, which the decoder will then output
a jibberish image. Second, the autoencoder is driven by reconstruction only, and therefore it creates a latent space that is very irregular, 
such that each of the digits is mapped to a very small "island" and diverging from it even a bit could result in a nonsense vector.
So the sulf-supervised autoencoder is practically useless for image generation and Bob is wrong. <br/> <br/>
There are multiple differences between "standard" autoencoders to VAEs:
Autoencoders encodes each image to a fixed point. VAEs output a distribution, usually normal distribution, then a latent vector
is sampled out of that distribution. <br/>
Autoencoders train for reconstruction objective only. VAEs train for reconstruction but also has a regularization term
in a shape of KL-Divergence that measures how much the latent space is similar to the normal distribution. <br/>
This training objective enforces the latent space to be structured and smooth in order to have meaningful random latent vectors 
in oppose to the irregular latent space of the autoencoders. <br/><br/>
Now let's see why this helps VAEs to act as generative models: <br/>
As written, VAEs output a distribution of the latent space. Each area of that distribution corresponds with a certain digit,
or at least a "mix of digits". Therefore, sampling out of that distribtuion a latent vector will be decoded as a meaningful image. <br/>
Moreover, in that latent space we could find other spatial features of the space other than the classes, such thickness,
curlyness, slantness and so on, so if we would learn the space generated we could control how will our images will turn out
by simply sampling from the "right" area in the distribution or even doing "latent space arithematic" between different vectors that have certain features. All of that thanks to training objective and the model's architecture.
</p>
