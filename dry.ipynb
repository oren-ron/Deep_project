### Question 1

<p>Increasing number of dimensions of h can initially increase the accuracy of the classification task, 
because it will give the autoencoder more capacity to encapsulate more data in each latent image.
But from a certain point (roughly around the number of dimensions of the image itself), the model can start learning the identity mapping
and therefore the classification task will have no benefit of the encoding, or worse, the model can start overfitting and memorizing the train data,
which will obviously result in worse results the more h will be increased.</p>

### Question 2

<p>The UAT says that a 1-layer MLP with a finite number of neurons can approximate fully any continious function working on finite spaces.
Therefore if we would use a big enough 1-layer MLP we can get perfect accuracy without the need of numerous layers in order
to achieve optimal error over any dataset and loss function.<br/><br/>
The conculsion is wrong because it is impractical. The layer size needed for perfectly classifying a dataset (such as CIFAR10) 
will be enormously large and therefore the training computation will never end. <br/>
We can get the same effect of 1 enormous layer with increasing the depth with much narrower layers. 
Increasing depth will decrease the wideness of the layers needed exponentially and that is much more pratical computations.
Also with deeper networks, we can introduce regularization techniques so that we won't overfit to the train set, and produce better generalization.
</p>

### Question 3

<p>Main advantage of CNNs over MLPs is the spatial structure that is CNNs take advantage of.
The convolution operation seek for a specific local pattern all around the image and that is benefitial because 
you can extract simple features such as shapes and gradients and later on a more complex patterns such as eyes, wheels, legs and so on.
MLPs searches for absolute spaced features so even if the model will recognize a face for example, cropping the image and feeding it again will cause the model to fail, because the face is now located somewhere else in the image. <br/><br/>
Even though both operations are linear there is a big difference. 
CNNs use the same kernel for the whole image so the number of weights is significantly smaller, which help us train the model more efficiently,
and reduce overfitting in oppose to MLP.<br/>
As written before, CNNs utilize the local patterns of images in a growing receptive field in order to make a guess,
aggragating the spatial information of the image into one complete picutre, while MLP cannot do it in its nature.<br/>
</p>

<p> Using linear-averaging instead of EMA will surely comparmise the algortihm's efficiency.
EMA is much more affected off the recent gradients than the linear one, which gives equal weight to all gradients.
Therefore if our algorithm "wants to make a turn", the EMA will do so much faster then the linear one.
Also, at some point the linear one will gather so much gradients that it won't be changed by adding another gradient because its proportion will be too small,
therefore we could get stuck.
